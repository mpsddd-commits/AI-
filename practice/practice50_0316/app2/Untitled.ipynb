```python
from sqlalchemy import create_engine
engine_mariadb = create_engine('mysql+pymysql://root:1234@192.168.0.107:23306/edu')
```

```python
from sqlalchemy import inspect
inspector = inspect(engine_mariadb)
tables = inspector.get_table_names()
print(tables)
```

```python
import pandas as pd
result = pd.read_sql_query("select * from edu.melon", engine_mariadb)
# print(result)
```

```python
result.comlumns = ["code", "img", "title"]
result.head()
```

```python
result.title
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("study2").master("spark://master:7077").getOrCreate()
```

```python
sDf = spark.createDataFrame(result)
```

```python
sDf.coalesce(1).write.format("study2.csv").option("header","true").save(path="/opt/spark/data/study2.csv",format="csv",mode="overwrite")
```
